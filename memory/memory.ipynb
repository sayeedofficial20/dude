```python
from langchain_core.runnables import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
```

```python
mssg_history = ChatMessageHistory()
```

```python
mssg_history.add_user_message('Hello,I am sayeed')
mssg_history.add_ai_message('Hello, sayeed! How can I assist you today?')
```

```python
mssg_history
print(mssg_history.messages)
print(mssg_history)
```

```python
store={}
```

```python
def get_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]    
```

```python
llm=ChatOpenAI(
    model_name="gpt-4o",
    temperature=0)
prompt=ChatPromptTemplate.from_messages(
    [("system","You are a helpful assistant."),
     ("human","{input}")]
)
parser=StrOutputParser()
chain=prompt|llm|parser
```

```python
chain_with_history = RunnableWithMessageHistory(chain,get_history,input_message_key="input",output_message_key="output")
```

```python
result=chain_with_history.invoke("whats my name?",config={"session_id": "sayeed123"})
print(result)
```

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_groq import ChatGroq
```

```python
# llm = ChatGroq(model="allam-2-7b", temperature=0)
llm = ChatGroq(model="openai/gpt-oss-20b", temperature=0)
```

```python
prompt_v1=ChatPromptTemplate.from_messages(
    [
        ('system','You are a friemdly chatbot.You have the access to past conversation history with the user. Alwaays use it to continue the conversation.'),
        MessagesPlaceholder(variable_name='history'),
        ('human','{input}')
    ]
)
```

```python
store_new={}
```

```python
def get_history_v2(session_id: str):
    if session_id not in store_new:
        store_new[session_id]= ChatMessageHistory()
    return store_new[session_id]
```

```python
parser=StrOutputParser()
```

```python
chain_c1=prompt_v1|llm|parser
```

```python
chain_with_history_v2=RunnableWithMessageHistory(chain_c1,get_history_v2,input_messages_key="input",output_messages_key="output",history_messages_key="history")
```

```python
result = chain_with_history_v2.invoke(
    {"input": "who am i?"},
    config={"configurable": {"session_id": "sayeed126"}},
)

print(result)
```
