```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
load_dotenv()
```

```python
from groq import Groq
import os

api_key = os.getenv("GROQ_API_KEY")

client = Groq(api_key=api_key)

models = client.models.list()

for model in models.data:
    print(model.id)
```

```python
llm=ChatGroq(model='allam-2-7b')
```

```python
parser=StrOutputParser()
```

```python
# prompt_template=PromptTemplate(template='Translate the following English text to Telugu: {text}', input_variables=['text'])
prompt_template=PromptTemplate(template='You are an very intelligent Translation agent. Please translate the provided {input} purely to {lang} language.Strictly output Must be in {lang}.Dont add any extra text otherthan the translation.', input_variables=['input', 'lang'])
```

```python
chainv1 = prompt_template | llm | parser
```

```python
result_v1=chainv1.invoke({'input':'Hello, how are you?', 'lang': 'Telugu'})
print(result_v1)
```
