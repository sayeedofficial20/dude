```python
def detect_high_readings(values, threshold):
    alerts = []

    for value in values:
        if value > threshold:
            alerts.append(value)

    return alerts


readings = [10, 12, 11, 45, 13, 52, 12]
print(detect_high_readings(readings, 40))
```

```python
import pandas as pd

df = pd.DataFrame({
    "machine": ["P1", "P1", "P2", "P2"],
    "temperature": [30, 32, 35, 55],
    "flow_rate": [100, 95, 90, 60]
})

print(df.groupby("machine")["temperature"].mean())

df["high_temperature"] = df["temperature"] > 40

print(df[df["high_temperature"]])
```
