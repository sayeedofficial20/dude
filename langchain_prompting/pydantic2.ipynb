```python
from pydantic import BaseModel,Field
from typing import List,Optional,Literal
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
```

```python
class Data(BaseModel):
    Product: str =Field(description="Name of the product")
    pos: List[str] =Field(description="List of positive reviews received for the product")
    nav: List[str] =Field(description="List of negative reviews received for the product")
    Sentiment: Literal['Positive','Negative','Neutral'] =Field(description="Overall sentiment of the product based on the reviews")
    rating: float=Field(ge=0, le=5, description="Rating provided by the user and should be between 0 and 5 as per review")
    reviewer: Optional[str] =Field(description="Name of the reviewer, if provided",default=None)
```

```python
review="""This smartphone is an outstanding device that perfectly balances performance, aesthetics, and everyday utility. The vibrant AMOLED display makes streaming content and navigating apps incredibly smooth and immersive. Performance is lightning-fast, handling intensive multitasking and gaming effortlessly without any noticeable lag or thermal throttling. I am highly impressed by the battery endurance, which comfortably clears a full day of heavy use, paired with incredibly rapid fast-charging capabilities. The camera system captures stunningly sharp images with rich color accuracy, even in challenging low-light environments. Offering premium features at a competitive price point, it delivers phenomenal value and comes highly recommended."""
```

```python
chatmodel=ChatOpenAI(model="gpt-4o-mini")
```

```python
struct_model=chatmodel.with_structured_output(Data)
```

```python
result=struct_model.invoke(review)
```
