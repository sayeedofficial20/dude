```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnableParallel
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt_temp1=PromptTemplate(template='Genearate a summary on this {topic}',input_variables=['topic'])
```

```python
prompt_temp2=PromptTemplate(template='Generate a Quiz question on this {topic}',input_variables=['topic'])
```

```python
gemini_model=ChatGoogleGenerativeAI(model="gemini-3.8-flash")
openai_model=ChatOpenAI(model='gpt-4o-mini')
```

```python
parser=StrOutputParser()
```

```python
parallel_chain=RunnableParallel({'summary':prompt_temp1 | gemini_model| parser,
                  'quiz': prompt_temp2 | openai_model| parser})
```

```python
prompt_temp3=PromptTemplate(template='Merge the following {summary} and {quiz} together',input_variables=['summary','quiz'])
```

```python
merge_chain=prompt_temp3 | openai_model|parser
```

```python
final_chain= parallel_chain | merge_chain
```

```python
result=final_chain.invoke({'topic' : 'Agentic AI'})
```
