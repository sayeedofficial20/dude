```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
google_model=ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)
```

```python
openai_model=ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
prompt_template1 = PromptTemplate(template="Generate the Detailed Report on the Topic:{topic}",input_variables=["topic"])
```

```python
prompt_template2 = PromptTemplate(template="Create a 7 bullet points summary on Report:{report}",input_variables=["report"])
```

```python
parser = StrOutputParser()
```

```python
chain=prompt_template1 | google_model | parser | prompt_template2 | openai_model | parser
```

```python
response=chain.invoke({"topic": "Artificial Intelligence Vs Machine Learning"})
```

```python
print(response)
```
