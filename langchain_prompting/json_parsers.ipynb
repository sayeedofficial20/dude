```python
# model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
# model_gog=ChatGoogleGenerativeAI(model="gemini-3.5-flash")
model_gog=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
openai_model = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
parser = JsonOutputParser()
```

```python
prompt_template=PromptTemplate(template='provide me with the Name,Age,Gender,City of a ficitional person and {specific_format}',input_variables=[],partial_variables={'specific_format':parser.get_format_instructions()})
```

```python
final_prompt=prompt_template.format()
```

```python
# response0=model.invoke(final_prompt)
response1=model_gog.invoke(final_prompt)
```

```python
response=openai_model.invoke(final_prompt)
```

```python
# print(response0.content)
print(response1.content)
print(response.content)
```

```python
prompt_sample=PromptTemplate(template="Give me any 7 Asia countries and its capitals in the {specific_format}",input_variables=[],partial_variables={'specific_format':parser.get_format_instructions()})
```

```python
chain_new=prompt_sample | openai_model | parser
```
