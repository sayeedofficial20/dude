```python
from langchain_core.runnables import RunnableSequence
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt_temp=PromptTemplate(template='Generate the Joke on this {topic}',input_variables=['topic'])
```

```python
llm=ChatOpenAI(model='gpt-4o-mini',temperature=0.5)
```

```python
parser=StrOutputParser()
```

```python
chainOne=RunnableSequence(prompt_temp,llm,parser)
```

```python
res=chainOne.invoke({'topic':'Gen AI'})
res
```

#applying Passthrough

```python
from langchain_core.runnables import RunnablePassthrough,RunnableParallel
```

```python
prompt_v1=PromptTemplate(template='Generate a joke on this {topic}',input_variables=['topic'])
```

```python
chain_one=RunnableSequence(prompt_v1,llm,parser)
```

```python
prompt_v2=PromptTemplate(template='Generate the Explanation for the following {joke}',input_variables=['joke'])
```

```python
explain_chain=prompt_v1 |llm|parser|prompt_v2|llm|parser
```

```python
explain_chain.invoke({'topic':'AI can replace human jobs'})
```

```python
parallel=RunnableParallel({'Joke': RunnablePassthrough(),
                           'Explanation':prompt_v2|llm|parser})
```

```python
final_chain=chain_one|parallel
```

```python
result=final_chain.invoke({'topic':'AI can replace human jobs'})
```
