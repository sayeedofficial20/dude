```python
from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.prompts import PromptTemplate
from pydantic import BaseModel, Field
from typing import List
```

```python
from langchain_openai import ChatOpenAI
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
class Data(BaseModel):
    name:str=Field(description='return the name of the user')
    age:int=Field(ge=21,description='return the age of thr user')
    city:str=Field(description='the city must to be of Telangana ONLY')
    state:str=Field(description='the state must be Telangana')

class Info(BaseModel):
    people: List[Data] = Field(
        description="List containing any number of fictional people Detais"
    )
```

```python
parser_real=PydanticOutputParser(pydantic_object=Info)
```

```python
prompt_sample=PromptTemplate(template='Provide the Details of any {count} ficitional persons which includes Name,age,city and state of  from {nationality} in {format}',input_variables=['nationality','count'],partial_variables={'format':parser_real.get_format_instructions})
```

```python
openai_model=ChatOpenAI(model='gpt-4o-mini')
```

```python
model_gog=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite",temperature=0.5)
```

```python
# chain=prompt_sample| openai_model | parser_real
chain_gemini=prompt_sample| model_gog | parser_real
```

```python
# response=chain.invoke({'nationality':'India'})
response_gemini=chain_gemini.invoke({'nationality':'India','count':10})
```
