```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_groq import ChatGroq
```

```python
prompt_sample=PromptTemplate(template='create a {count} points Summary on {topic}',input_variables=['count','topic'])
```

```python
# model=ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)
model=ChatGroq(model="openai/gpt-oss-120b", temperature=0)
```

```python
parser=StrOutputParser()
```

```python
chain=prompt_sample | model
```

```python
res=chain.invoke({'count':5,'topic':'AI in Healthcare'})
print(res)
```
