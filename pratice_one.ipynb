```python
documents_new=["Shah Rukh Khan is known as the King of Bollywood and has acted in films like Dilwale Dulhania Le Jayenge.", "Rajinikanth is a superstar from South Indian cinema and is famous for his unique style and dialogues.", "Amitabh Bachchan is one of the greatest actors in India and has worked in classics like Sholay and Don."]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from dotenv import load_dotenv
```

```python
embeddings_v1=OpenAIEmbeddings(model="text-embedding-3-large", dimensions=69)
```

```python
query_q1="Who is the King of Bollywood?"
```

```python
query_results=embeddings_v1.embed_query(query_q1)
doc_results=embeddings_v1.embed_documents(documents_new)
```

```python
final_results=cosine_similarity([query_results], doc_results)[0]
```

```python
print(final_results)
```

```python
scores=sorted(list(enumerate(final_results)), key=lambda x: x[1], reverse=True)[0]
print(scores)
```

```python
print("Question:", query_q1)
print("Answer:", documents_new[scores[0]])
print("Similarity Score:", scores[1])
```

```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
query="shifting from Deep Learning to GenAI"
```

```python
retriever=WikipediaRetriever(top_k_results=3,lang='en')
```

```python
answer1=retriever.invoke(query)
```
