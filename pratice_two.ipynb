FAISS

```python
from langchain_community.vectorstores import FAISS

from langchain_core.documents  import Document
```

```python
questions = [
"Sayeed is a bad boy",
"Ganesh is very very good boy",
"Singh acts like a good boy.but not good boy",
"pavan is a bad boy",
]
```

```python
query="List the bad boys from the above questions."
```

```python
str_obj=FAISS.from_texts(questions,embeddings)
```

```python
new_retiever=str_obj.as_retriever(search_type='similarity',search_kwargs={"k": 2, "lambda_mult": 1})
```

```python
results=new_retiever.invoke(query)
results
```

```python
from langchain_classic.retrievers import MultiQueryRetriever
from langchain_openai import OpenAIEmbeddings,ChatOpenAI
from langchain_community.vectorstores import FAISS
```

```python
embeddings_mqr=OpenAIEmbeddings()
chat_obj=ChatOpenAI(model='gpt-3.5-turbo',temperature=0)
```

```python
mqr_query='List the good boys from the given data whose age were greater than 27.'
```

```python
mqr_docs = [
"Sayeed is a bad boy.and age is 24",
"Ganesh is very very good boy.and age is 27",
"Singh acts like a good boy.but not good boy and age is 35",
"pavan is a good boy and age is 26",
"abhi is a good boy and age is 30"
]
```

```python
mqr_vect_store=FAISS.from_texts(mqr_docs,embeddings_mqr)
```

```python
multi_retiever=MultiQueryRetriever.from_llm(retriever=mqr_vect_store.as_retriever(),llm=chat_obj)
```

```python
output=multi_retiever.invoke(query)
```

```python
for i in output:
    print(i.page_content)
```

```python
from langchain_classic.retrievers import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
```

```python
query='How to reset my Router password ?'
```

```python
docs = ["""
Routers connect multiple devices to a network and manage communication between them.
Most home routers provide features such as Wi-Fi, DHCP, firewall settings, guest networks, and parental controls.
Routers commonly operate on 2.4 GHz and 5 GHz wireless bands, with different ranges and speeds.
The router administration page allows users to configure network settings, security options, firmware updates, and connected devices.
Router passwords can be changed from the wireless security settings in the administration interface.
Some routers also support Ethernet connections, VPN configuration, port forwarding, and quality-of-service settings.
"""]
```

```python
ccr_vs=FAISS.from_texts(mqr_docs,embeddings)
```

```python
base_retriever=ccr_vs.as_retriever()
```

```python
compressor=LLMChainExtractor.from_llm(chat_obj)
```

```python
ccr_retriever=ContextualCompressionRetriever(base_compressor=compressor,base_retriever=base_retriever)
```

```python
results_ccr=ccr_retriever.invoke(mqr_query)
```

```python
results_ccr
```

```python
for i in results_ccr:
    print(i.page_content.strip())
```
